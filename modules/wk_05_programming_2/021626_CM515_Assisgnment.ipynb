## 2. Make a function called `revcomp()` that prompts the user for a DNA sequence and returns the reverse complement. Have the function inform the user if they put RNA instead.

```python
def revcomp():
    # Calculates the reverse complement of a DNA sequence
    # If the input contains 'U', then it is RNA and the function will inform the user and exit
    comp_dict = {"A": "T", "T": "A", "C": "G", "G": "C"}
    dna_seq = input("enter a DNA sequence: ")
    uppercase_dna_seq = dna_seq.upper()
    comp_seq = ""
    if "U" in uppercase_dna_seq:
        print("RNA sequence entered, please enter a DNA sequence")
        return  # Exit the function if RNA is detected
    for base in uppercase_dna_seq:
        comp_seq += comp_dict[base]
        rev_comp_seq = comp_seq[::-1]
    return f"reverse complement of {uppercase_dna_seq} is {rev_comp_seq}"

revcomp()
```

Remake the code into  a function called `savings_calc()` that prompts the user for the same inputs as before, `principle, interest, and years`.

This time, the function should give you the result of compounding interest (whatever your total is after 1 year, should become the principal for the next year, and so on for however many years the function is designed to calculate interest for).

    return f"(You started with ${principle} in your savings account which has {interest} annual interest. You ended up with ${final_savings} after investing for {years} years."

```python
def savings_calc():
    # Calculates the final amount in a savings account
    # Takes as input the initial amount of money, the annual interest rate, and the number of years
    # Compounds the interest annually and returns the final amount
    principle = float(input("Enter the initial amount of money: "))
    interest = float(input("Enter the annual interest rate (as a percentage): "))
    rate = interest / 100
    years = float(input("Enter the number of years: "))
    year_counter = 0
    final_savings = principle
    while year_counter < years:
        final_savings = final_savings * (1 + rate)
        year_counter += 1
    return f"You started with ${principle} in your savings account that has %{interest} annual interest. You ended up with ${final_savings:.2f} after investing for {years} years."

savings_calc()
```

# 5. Try making a function called `genre_sorter()`.

This function should:

Take the dictionary below of movies/shows with associated genres and return a summary that looks like this:

Here is the selection:  
        Comedies: ...
        Horror: ..., ..., ...
        Romance: ...
        Drama: ..., ...
        Action/Adventure: ...
        Kids: ...

You want to set up your function such that you will be able to do further work with these movie lists at another time (not just have them in your function).


*Apparently running this in a jupyter notebook will look very ugly because in a notebook or shell, a return string will show the \n. To see the cleaned up version of your function working, use: `print(genre_sorter(netflix_selection))`*

```python
netflix_selection = {"Bridgerton" : "Romance", "Naruto" : 'Anime', "Scream7" : "Horror", "The Land Before Time" : "Kids", 
                     "Hundred Beavers" : "Comedy", "The Notebook" : "Romance", "Phantom Thread" : "Drama", 
                     "The King's Speech" : "Drama", "One Piece" : "Anime", "Jurassic Park" : "Action/Adventure", 
                     "Silent Hill" : "Horror", "Pet Semetary" : "Horror", "Mission Impossible" : "Action/Adventure"}

def genre_sorter(netflix_selection):
    # Sorts the netflix selection dictionary by genre
    # Returns a new dictionary with the genres as keys and the movie titles as values
    sorted_netflix_selection = {}
    for title, genre in netflix_selection.items():
        sorted_netflix_selection.setdefault(genre, []).append(title)
    return f" Here is the selection: {sorted_netflix_selection}"

genre_sorter(netflix_selection)
```

# 6. Try making a function called `average()`.

This function should:

    Prompt the user for input continuously.
        
    Be able to provide the user with an average of numbers when they are done providing numbers (by having the user type someting like 'stop', or 'finished').

    Give the user an option to exit out of the program.
        
    Be able to address common pitfalls a user might have (There are two big ones).


```python
def average():
    # Calculates the mean of a list of numbers
    # Takes as input a list of numbers from the user until they enter 'stop'
    # If the user enters something that is not a number or 'stop', it will inform the user and ask for input again
    # If the user enters 'stop' before entering any numbers, it will inform the user that no numbers were entered
    # Once the user enters 'stop', it will calculate and return the mean of the numbers entered
    numbers = []
    while True:
        number = input("Enter a number (or 'stop' to finish):")
        if number.lower() == 'stop':
            break
        try:
            numbers.append(float(number))
        except ValueError:
            print("Invalid input, please enter a number or 'stop'")
    if len(numbers) == 0:
        return "No numbers entered."    
    mean = sum(numbers) / len(numbers)
    return f"mean of {numbers} is {mean}"

average()
```
